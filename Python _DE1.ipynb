```python
#Problem 1: Filter and Transform (The Clean-Up)
# * Context: You receive a raw stream of event metrics (numbers). Some are corrupt (represented as negative numbers or strings), and some need normalization.
 #* Task: Given a mixed list of integers and strings, write a snippet that:
 #  * Filters out any non-integer values or negative numbers.
  # * Multiplies the remaining valid integers by 10 (simulating a scaling factor).
 #* Example: [5, "error", -3, 12, "42", 2] \rightarrow Output: [50, 120, 20]

data= [5, "error", -3, 12, "42", 2]
result = [x*10 for x in data if isinstance(x,int) and not isinstance(x,bool) and x>=0]
print(result)
```

```python
#Problem 2: Flattening Nested Arrays
data=[1, [2, 3], [4, [5, 6]], 7]
result=[]
def flatten(data):
 for item in data:
     if isinstance(item,list):
         flatten(item)
     else:
         result.append(item)
flatten(data)
print(result)
```

```python
#Problem 3: Length-Based String Filtering (Token Cleaning)
data=["A", "Data", "I", "Pipeline", "X"]
result=[word.lower() for word in data if len(word)>1]
print(result)
```

```python
#Problem 4: Aggregating by Category (Group-By Logic)
data = [
    {"category": "cloud", "amount": 100},
    {"category": "database", "amount": 50},
    {"category": "cloud", "amount": 200}
]
result={}
for item in data:
    category=item["category"]
    amount=item["amount"]
    if category in result:
        result[category]+=amount
    else:
        result[category]=amount
print(result)
```

```python
#Problem 5: Handling Missing Keys (Defensive Data Parsing)

users = [
    {"id": 1, "status": "active"},
    {"id": 2},
    {"id": 3, "status": "inactive"}
]

result = {
    "active": 0,
    "inactive": 0,
    "unknown": 0
}

for user in users:
    status = user.get("status", "unknown")

    if status not in result:
        status = "unknown"

    result[status] += 1

print(result)
      
        
    
```

```python
#Problem 6: Flattening a Nested Configuration Dictionary

data = {
    "db": {
        "host": "aws",
        "port": 3306
    }
}

result = {}

def flatten_dict(data, prefix=""):
    for key, value in data.items():
        new_key = f"{prefix}.{key}" if prefix else key

        if isinstance(value, dict):
            flatten_dict(value, new_key)
        else:
            result[new_key] = value

flatten_dict(data)

print(result)
    
```
